### Multimodal example

```python
from transformers import BlipProcessor, BlipForQuestionAnswering
from PIL import Image

processor = BlipProcessor.from_pretrained("Salesforce/blip-vqa-capfilt-large")
model = BlipForQuestionAnswering.from_pretrained("Salesforce/blip-vqa-capfilt-large")

image = Image.open("/content/beautiful-shot-white-british-shorthair-kitten.jpg").convert("RGB") # supply a valid image path, convert to RGB
question = "Is there an elephant?"

# Prepare inputs for VQA
inputs = processor(image, question, return_tensors="pt")

# Generate answers
outputs = model.generate(**inputs)
answer = processor.decode(outputs[0], skip_special_tokens=True)
print(answer)
```

Gradio app!

```python
from transformers import BlipProcessor, BlipForQuestionAnswering
from PIL import Image
import gradio as gr

# Load BLIP VQA model
processor = BlipProcessor.from_pretrained("Salesforce/blip-vqa-capfilt-large")
model = BlipForQuestionAnswering.from_pretrained("Salesforce/blip-vqa-capfilt-large")

def answer_question(image, question):
    if image is None or question.strip() == "":
        return "Please provide an image and a question."

    # Ensure image is in RGB format
    image = image.convert("RGB")

    # Prepare inputs for VQA
    inputs = processor(image, question, return_tensors="pt")

    # Generate answers
    outputs = model.generate(**inputs)
    answer = processor.decode(outputs[0], skip_special_tokens=True)
    return answer

with gr.Blocks() as demo:
    gr.Markdown("#  Visual Question Answering with BLIP")
    with gr.Row():
        with gr.Column(scale=1):
            image_input = gr.Image(type="pil", label="Upload an Image")
            question_input = gr.Textbox(label="Enter your question")
            with gr.Row():
                submit_btn = gr.Button("Get Answer", variant="primary")
                clear_btn = gr.Button("Clear", variant="secondary")

            # Add sample image loader
            gr.Examples(
                examples=["/content/beautiful-shot-white-british-shorthair-kitten.jpg"],
                inputs=image_input,
                label="Load Sample Image"
            )
        with gr.Column(scale=1):
            output_text = gr.Textbox(label="Answer", interactive=False)

    submit_btn.click(fn=answer_question,
                     inputs=[image_input, question_input],
                     outputs=output_text)

    clear_btn.click(fn=lambda: [None, "", ""], # Return None for image, empty string for question and output
                    inputs=[],
                    outputs=[image_input, question_input, output_text])

demo.launch()
```
